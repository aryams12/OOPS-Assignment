1. What is Object-Oriented Programming (OOP)?

    OOP is a programming paradigm based on the concept of objects, which contain data (attributes) and methods (functions). It focuses on organizing code around real-world entities rather than functions and logic.

2. What is a class in OOP?

    A class is a blueprint or template for creating objects. It defines attributes (data) and methods (functions) that describe the behavior of the objects.


3. What is an object in OOP?
    
    An object is an instance of a class. It represents a specific example of the class and has its own values for the class attributes.



4. What is the difference between abstraction and encapsulation?

    Abstraction: Hides complex implementation details and shows only the essential features.

    Encapsulation: Binds data and methods into a single unit (class) and restricts direct access using access modifiers.

5. What are dunder methods in Python?

    “Dunder” means “double underscore.”
    These are special methods like __init__, __str__, and __add__ used to define object behavior (also called magic methods).

6. Explain the concept of inheritance in OOP.


    Inheritance allows a class to acquire the properties and methods of another class.
    The base class is called the parent (superclass), and the derived class is the child (subclass).

7. What is polymorphism in OOP?

    Polymorphism means “many forms.” It allows one interface to be used for different data types — for example, methods with the same name but different implementations in subclasses.

8. How is encapsulation achieved in Python?


    Encapsulation is achieved using:

    Private variables (prefix _ or __)

    Getter and setter methods to control access

9. What is a constructor in Python?

    A constructor is a special method named __init__() that automatically runs when an object is created, initializing its attributes.

10. What are class and static methods in Python?


    Class method: Defined using @classmethod; works with the class itself (cls).

    Static method: Defined using @staticmethod; does not access class or instance data.

11. What is method overloading in Python?

    Python doesn’t support traditional overloading.
    Instead, you can use default arguments or variable-length arguments to simulate method overloading.

12. What is method overriding in OOP?

    Overriding occurs when a subclass provides a new implementation for a method already defined in the parent class.


13. What is a property decorator in Python?

    The @property decorator turns a method into a getter, allowing you to access it like an attribute (e.g., obj.name instead of obj.get_name()).

14. Why is polymorphism important in OOP?

    It enhances flexibility and reusability by allowing a single interface to represent different data types or behaviors (e.g., multiple classes implementing a speak() method differently).

15. What is an abstract class in Python?

    An abstract class is a class that cannot be instantiated directly and may contain abstract methods — declared but not implemented.
    Defined using the abc module.

16. What are the advantages of OOP?


    Code reusability

    Better organization and modularity

    Easier maintenance and debugging

    Data security through encapsulation

    Scalability via inheritance and polymorphism

17. What is the difference between a class variable and an instance variable?

    Class variable: Shared by all objects of a class.

    Instance variable: Unique to each object.

18. What is multiple inheritance in Python?

    Multiple inheritance allows a class to inherit from more than one parent class.

    class A: pass

    class B: pass

    class C(A, B): pass

19. Explain the purpose of __str__ and __repr__ methods in Python.

    __str__: Returns a readable, user-friendly string representation.

    __repr__: Returns a detailed, developer-friendly representation (used in debugging).

20. What is the significance of the super() function in Python?

    super() calls methods from the parent class, often used to invoke the parent constructor (__init__) or overridden methods.

21. What is the significance of the __del__() method in Python?
   
    
    __del__() is a destructor method that is called when an object is deleted or goes out of scope, used for cleanup operations.

23. How does polymorphism work in Python with inheritance?

    If a subclass overrides a parent method, Python will use the subclass version when called on an instance of the subclass.

24. What is method chaining in Python OOP?

    Method chaining means calling multiple methods sequentially in a single line, each returning self.


    class Demo:

    def step1(self): print("Step 1"); return self
    
    def step2(self): print("Step 2"); return self

    obj = Demo().step1().step2()

25. What is the purpose of the __call__() method in Python?

    When defined in a class, it allows an instance to be called like a function.


    class Greet:

    def __call__(self, name):
    
        print("Hello", name)

    obj = Greet()

    obj("Aryaman")  # behaves like a function
